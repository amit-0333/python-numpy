###`Q-1:` Find the nearest element in the array to a given integer.
```
Ex:-
a=23 and array - [10 17 24 31 38 45 52 59].
Nearest element is 24
```
Hint: Read about this function `argmin()`

````python
# APPROACH:
# Find absolute difference between each array element and input number.
# The minimum difference element is the closest number.

# TC: O(n)
# SC: O(n)

import numpy as np

a = int(input('Enter number: '))

array = np.array([10, 17, 24, 31, 38, 45, 52, 59])

# Find index of closest element
idx = np.abs(array - a).argmin()

print("Closest number:", array[idx])
````

###`Q-2:` Replace multiples of 3 or 5 as 0 in the given array.
```
arr=[1 2 3 4 5 6 7 9]

result-> [1 2 0 4 0 0 7 0]
```

````python
# code here
# APPROACH:
# Use np.where() to check condition.
# If element is divisible by 3 or 5, replace with 0.
# Otherwise keep original value.

# TC: O(n)
# SC: O(n)

import numpy as np

arr = np.array([1, 3, 5, 7, 9, 10, 12, 14])

result = np.where((arr % 3 == 0) | (arr % 5 == 0), 0, arr)

print(result)
````

###`Q-4:` Your are given an array which is havig some nan value. You job is to fill those nan values with most common element in the array.
```
arr=np.array([[1,2,np.nan],[4,2,6],[np.nan,np.nan,5]])

```

````python
# code here
# APPROACH:
# Find mean excluding NaN values using np.nanmean().
# Replace NaN positions with that mean.

# TC: O(n)
# SC: O(1)

import numpy as np

arr = np.array([
    [1, 2, np.nan],
    [4, 2, 6],
    [np.nan, np.nan, 5]
])

# Mean excluding NaN
mean = np.nanmean(arr)

# Replace NaN with mean
arr[np.isnan(arr)] = mean

print(arr)
````

### `Q-5:` Write a NumPy program

- to find the missing data in a given array. Return a boolean matrix.
- also try to fill those missing values with 0. For that, you can use `np.nan_to_num(a)`

```python
import numpy as np

np.array([[3, 2, np.nan, 1],
          [10, 12, 10, 9],
          [5, np.nan, 1, np.nan]])
```

````python
# code here
# APPROACH:
# Use np.isnan() to find missing values (NaN).
# It returns a boolean matrix.
# Then use np.nan_to_num() to replace NaN with 0.

# TC: O(n)
# SC: O(n)

import numpy as np

arr = np.array([
    [1, 2, np.nan],
    [4, np.nan, 6],
    [7, 8, np.nan]
])

# Boolean matrix for missing values
missing = np.isnan(arr)

print("Missing value matrix:")
print(missing)

# Replace NaN with 0
filled = np.nan_to_num(arr)

print("\nArray after filling NaN with 0:")
print(filled)
````

###`Q-6:` Given two arrays, X and Y, construct the Cauchy matrix C.
`Cij =1/(xi - yj)`

http://en.wikipedia.org/wiki/Cauchy_matrix
```
x = numpy.array([1,2,3,4]).reshape((-1, 1)
y = numpy.array([5,6,7])
```

````python
# code here
# APPROACH:
# Use broadcasting.
# Each element is calculated as:
# Cij = 1 / (xi - yj)

# TC: O(m*n)
# SC: O(m*n)

import numpy as np

x = np.array([1, 2, 3, 4]).reshape((-1, 1))

y = np.array([5, 6, 7])

# Construct Cauchy matrix
C = 1 / (x - y)

print(C)
````

###`Q-7:` Plot this below equation.

$$\large y = \frac{e^x - e^{-x}}{e^x + e^{-x}}$$

**Note: This equation is called tanh activation function. In deep learning, many times this function is used. If you find some difference between the sigmoid function and this tanh function, note that to your notebook.**

````python
# code here
# APPROACH:
# Generate x values using linspace.
# Apply tanh equation.
# Plot using matplotlib.

# TC: O(n)
# SC: O(n)

import numpy as np
import matplotlib.pyplot as plt

# Generate x values
x = np.linspace(-10, 10, 100)

# tanh activation function
y = (np.exp(x) - np.exp(-x)) / (np.exp(x) + np.exp(-x))

# Plot
plt.plot(x, y)

plt.title("Tanh Activation Function")
plt.xlabel("x")
plt.ylabel("y")

plt.grid(True)

plt.show()
````

### `Q-9:` Write a program implement Boradcasting Rule to check if two array can be added or not.
Given tuples of shapes.
```
shape of a- (3,2,2)
shape of b- (2,2)

check_broadcast(a, b) -> return Boolean (True if can broadcasted, False other wise.)
```

````python
# code here
# APPROACH:
# Compare dimensions from right to left.
# Dimensions are compatible if:
# 1. both are equal
# 2. one of them is 1
# Otherwise broadcasting is not possible.

# TC: O(max(n, m))
# SC: O(1)

def check_broadcast(a, b):

    # Reverse shapes for right-to-left comparison
    a = a[::-1]
    b = b[::-1]

    # Compare dimensions
    for i in range(max(len(a), len(b))):

        dim_a = a[i] if i < len(a) else 1
        dim_b = b[i] if i < len(b) else 1

        # Broadcasting rule
        if dim_a != dim_b and dim_a != 1 and dim_b != 1:
            return False

    return True


# Given shapes
a = (3, 2, 2)
b = (2, 2)

print(check_broadcast(a, b))
````
